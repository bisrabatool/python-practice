```python
import seaborn as sns
import matplotlib.pyplot as plt

null_values = df.isnull().sum()
print("Null Values in Each Column:\n", null_values)

plt.figure(figsize=(5, 3))
sns.heatmap(df.isnull(), cbar=True, cmap='Blues')
plt.title('Heatmap of Null Values in Dataset')
plt.show()
```

```python
import numpy as np
df.replace('?', np.nan, inplace=True)

print(df)
```

```python
print("Null Values in Each Column:\n", df.isnull().sum())
```

```python
duplicates = df.duplicated()
print("Duplicate Rows:\n", df[duplicates])
duplicate_count = duplicates.sum()
print(f"Number of duplicate rows: {duplicate_count}")
```

```python
df_cleaned = df.drop_duplicates()
print("Data after removing duplicates:\n", df_cleaned)
```

```python
statistics = df.describe()
print("all Statistics:\n", statistics)
```

```python
print("Original DataFrame:\n", df.head())
df_dropped = df.drop(columns=['education-num', 'capital-gain', 'capital-loss'])
print("\nDataFrame After Dropping Columns:\n", df_dropped.head())
```
